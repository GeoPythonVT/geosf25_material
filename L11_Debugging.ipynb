## Code 2 - Part 2
``` python
# Compute anomalies
df["var1_anom"] = df["var1"] - df["var1"].mean()
df["var2_anom"] = df["var2"] - df["var2"].mean() 

# Compute correlation of anomalies
corr = df["var1_anom"].corr(df["var2_anom"])
print("Correlation:", corr)
```

## Exercise 2
1. Execute part 1 of the code above in a code cell. This should not raise an error.
2. Exectute part 2, this should raise an error. What error is raised and in which line?
4. Enter only part 2 of the code in a LLM web-app (e.g., ChatGPT, Google Gemini, Microsoft Copilot, ...). (Note, this is more realistic, as often you read data from an external source and AI cannot inspect content of the variables.)
5. Do you get any insight?
6. Now do the same, but also provide the error name and the line the error is raised.
7. Narrow down the options, by providing more details on your output or problem context.


---
# Sources / Recommended Reading

The Python Debugging Tricks Nobody Taught Me: https://pub.towardsai.net/the-python-debugging-tricks-nobody-taught-me-7c894d099ff0

How to debug anything with AI: https://www.cyfrin.io/blog/ai-code-debugger-debug-code-with-ai

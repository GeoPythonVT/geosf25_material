#### Question 1: Random Forests

A Random Forest model is a machine learning algorithm that combines many individual decision trees to make predictions. It’s an example of an ensemble method, meaning it builds many “weak” models and merges them into a stronger one. Why does combining many simple decision trees usually lead to better predictions than using just one? And what is the purpose of adding randomness when building a random forest?

#### Question 2: Classification versus Regression

Using Random Forest models, one can achieve both classification and regression. Explain the difference between these modeling approaches. Name one example for each type of model.

#### Question 3: Learning from models

Random Forests can make very good predictions, but they are sometimes called “black-box” models. After trying them in the exercise, what parts of the model can you still understand or explain (for example, which features are most important)?
What parts are harder to understand? How might this influence how you would use Random Forests for real-world environmental or geoscience problems?

#### Copies vs. views

```python
# First, copies versus views:

m_view = m[2,:]       # does not create a new object
m_copy = m[2,:].copy() # creates a new (sliced) object

# Alter one of the elements and see what happens to m?!
```

```python
m_view[0] = 0
m # does change - be careful!
```

```python
m_copy[0] = -1
m # does not change
```

#### Rules of broadcasting

```python
# Rules of broadcasting:

twoDimArray = np.array([  [1, 2, 3],
                          [4, 5, 6]  ])
oneDimArray = np.array([100]) 
output = twoDimArray + oneDimArray
print("Broadcasting result:\n", output)

# Helps to keep code vectorized and prevents repetition of data
```

```python
# broadcasting does vectorize the following thing for you

numRow, numCol = twoDimArray.shape
output = twoDimArray.copy()
for row in range(numRow):
    for col in range(numCol):
        output[row][col] += oneDimArray[0] # this oneDimArray only has 1 element
print("For loop result:\n", output)
```

#### Filtering

```python
m[m>5] # use a boolean test test result to filter out values
```

```python
m[~(m>5)] # invert the filter
```

#### Filtering using a mask

When we filter data like this, a mask is created. The mask consists of True/False or 1/0 arrays:

```python
mask = v>5
mask
```

```python
v[mask]
```

- Masking/Filtering works for arrays of any dimensions
- It is also vectorized
- It can be applied with or without broadcasting:

```python
m
```

```python
mask = [True,True,False,False]
mask
```

```python
m[mask]   # broadcasted along rows (repeated for every row)
```

```python
m[:,mask] # applied along columns
```

```python
m[mask,:] # applied along rows
```

#### 3D array and 2D mask
Now let's do the same for a 3D and a 2D vector.
This can be helpful for spatiotemporal maps!

```python
# let's reuse the 3x4x5 tensor from earlier

mask2D = t[0,:,:]>4
mask2D
```

Keep in mind that matrices in tensor are axis 1 and 2 (not 0), e.g. for a spatiotemporal map, maps could be on axis 1 and 2 and the time would be on axis 1.

Then the mask can be applied at every time step:

```python
t[:,mask2D]
# The resulting three rows are filtered values from the map at each time step. 
```

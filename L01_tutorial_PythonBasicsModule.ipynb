### Modules vs. Scripts & Built-in `__name__`


What is the content of the built-in attribute `__name__` in a code cell ( = main program level)?


````python
__name__
````

What is the content of `__name__` in another module? 
To check, add `print(name) to the module, reimport the module. 

````python
# reimport the module
import importlib
importlib.reload(hello)
````

Based on the output you should now understand, that `__name__` has different values in the built-in and the global namescope.

• With Python command (shell or notebook) we are running code from the main program level!
    
    `__name__ = '__main__'`
    
• When importing, we are running code at import
    
    `__name__ = <modulename>`


Now, at the bottom of hello.py, define a code sequence that is called only, if the module is executed from the command line but not when the file is imported (as module). This can be achieved based on the content of the built-in variable `__name__` :

Add to the bottom of hello.py:
```python
if __name__ == '__main__':
    main()
```

Then move the print functions into a new function, called `main()`. Note, a main() function is not required, and you could give this function any other name. 

Now you can use the module either as script, useful to run the main function content, or as module, useful to reuse all the other packaged functions in this module file.

To run the module as script, either call the main function (in a jupyter notebook), or in a bash shell, call the module with the python command.
python hello


````python
! python hello.py   # via bash command, from any shell
````

````python
hello.main()       # run script content via main function call
````

To simply import the module, without running content in the main function, just import the module again. Now you will see that the content in the `main()` function is not executed.

````python
# reimport the module
import importlib
importlib.reload(hello)
````
